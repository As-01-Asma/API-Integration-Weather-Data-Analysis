## 02. Importing Required Libraries

```python
import requests
import pandas as pd
import json
```

## 03. Setting Up API Parameters

```python
params = {
    "latitude": 17.3850,
    "longitude": 78.4867,
    "hourly": [
        "temperature_2m",
        "relative_humidity_2m",
        "wind_speed_10m",
        "precipitation"
    ],
    "forecast_days": 1,
    "timezone": "Asia/Kolkata"
}
```

## 04. Setting Up Open-Meteo API Endpoint

```python
api_url = "https://api.open-meteo.com/v1/forecast"
```

## 05. Sending API Request

```python
response = requests.get(
    api_url,
    params=params,
    timeout=30
)
```

## 06. Checking API Response Status

```python
print("Status Code:", response.status_code)
```

## 07. Converting API Response to JSON

```python
data = response.json()
```

## 08. Checking Available API Response Keys

```python
print(data.keys())
```

## 09. Extracting Hourly Weather Data

```python
print(data["hourly"])
```

## 10. Creating Pandas DataFrame

```python
df = pd.DataFrame(data["hourly"])
```

## 11. Checking Dataset Shape

```python
print("Dataset Shape:", df.shape)
```

## 12. Checking Column Names

```python
print(df.columns)
```

## 13. Inspecting Dataset Information

```python
df.info()
```

## 14. Checking Missing Values and Duplicates

```python
print("Missing Values:")
print(df.isnull().sum())

print("Duplicate Records:")
print(df.duplicated().sum())
```

## 15. Generating Statistical Summary

```python
print(df.describe())
```

## 16. Converting Time Column to Datetime

```python
df["time"] = pd.to_datetime(df["time"])
```

## 17. Exporting Original API Response to JSON

```python
with open("api_response.json", "w") as file:
    json.dump(data, file, indent=4)
    
```

## 20. Importing Database Libraries

```python
import pyodbc
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
```

## 21. Checking Available ODBC Drivers

```python
print("Available ODBC Drivers:")
print(pyodbc.drivers())
```

## 22. Connecting to SQL Server Database

```python
connection_string = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=DESKTOP-JA5ED2F\\SQLEXPRESS;"
    "DATABASE=WeatherDB;"
    "Trusted_Connection=yes;"
)

connection_url = URL.create(
    "mssql+pyodbc",
    query={"odbc_connect": connection_string}
)

engine = create_engine(connection_url)

print("Successfully connected to SQL Server!")
```

## 23. Verifying SQL Server Connection

```python
with engine.connect() as connection:
    print("Database connection verified!")
```

## 24. Loading Weather Data into SQL Server

```python
df.to_sql("WeatherHourlyData", engine, if_exists="append", index=False)

print("Weather data successfully loaded into SQL Server!")
```

## 25. Verifying Data Loading into SQL Server

```python
print(pd.read_sql(
    "SELECT COUNT(*) FROM WeatherHourlyData",
    engine
))
```
